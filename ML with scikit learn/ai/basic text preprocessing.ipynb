```python
import nltk
nltk.download('punkt')
from  nltk.tokenize import word_tokenize

text='i am nilesh and i want to a data science'
tokens=word_tokenize(text)
print(tokens)
```

```python
# remove a stopwords
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words=set(stopwords.words('english'))

filtere=[word for word in tokens if word.lower() not in stop_words]
print(filtere)
```

```python
#stemming

from nltk.stem import PorterStemmer
stm=PorterStemmer()
stemmed=[stm.stem(word) for word in filtere]
print(stm)
```

```python
# lemmatizatiom

from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

lem=WordNetLemmatizer()
lemword=[lem.lemmatize(word) for word in filtere]
print(lem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
sen=['i am nilesh','data science']

vec=CountVectorizer()
x=vec.fit_transform(sen)

print(vec.get_feature_names_out())
print(x.toarray())
```

```python
# tf-idf

from sklearn.feature_extraction.text import TfidfVectorizer
vec=TfidfVectorizer()
x=vec.fit_transform(sen)
print(x.toarray())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

text=['i love nlp','i hate bugd','this is amazing','this is bed']
label=[1,0,1,0]

vec=CountVectorizer()
x=vec.fit_transform(text)

x_train,x_test,y_train,y_test=train_test_split(x,label,test_size=0.2)

model=LogisticRegression()
model.fit(x_train,y_train)

print(model.predict(vec.transform(['i an good'])))
```
