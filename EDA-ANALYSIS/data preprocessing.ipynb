# 1.data cleaning

```python
import pandas as pd
import numpy as np
from faker import Faker

fake = Faker()

# Generate 100 rows of fake customer data
data = {
    "Name": [fake.name() for _ in range(100)],
    "Age": np.random.randint(18, 70, 100),
    "JoinDate": [fake.date_between(start_date="-5y", end_date="today") for _ in range(100)],
    "Email": [fake.email() for _ in range(100)],
    "PurchaseAmount": np.random.choice([100, 200, None, 500, 1000], 100)  # includes missing values
}

df = pd.DataFrame(data)

# Add duplicates and inconsistencies
df.loc[5] = df.loc[10]   # duplicate row
df.loc[20, "Age"] = None # missing value
df.loc[30, "Email"] = "not_an_email" # invalid entry

print(df.head())
```
